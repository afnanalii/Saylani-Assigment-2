##### Question 3: Online Shopping Discount
An online store gives discounts based on the total purchase amount.
Ask the customer for the total amount.
- Less than 5,000 → No discount
- 5,000 to 9,999 → 10% discount
- 10,000 to 19,999 → 15% discount
- 20,000 or above → 20% discount

Calculate and display:
- Original amount
- Discount
- Final amount

```python
originalAmount = int(input("Tell your amount:- "))

if originalAmount >= 20000:
    discount = originalAmount * 20 / 100
    finalAmount = originalAmount - discount
elif originalAmount >= 10000:
    discount = originalAmount * 15 / 100
    finalAmount = originalAmount - discount
elif originalAmount >= 5000:
    discount = originalAmount * 10 / 100
    finalAmount = originalAmount - discount
else:
    discount = 0
    finalAmount = originalAmount

print(f"Original Amount:- {originalAmount}\n"
      f"Discount:- {discount}\n"
      f"Final Amount:- {finalAmount}")
```

##### Question 7: Number Analyzer
A data-entry application needs to analyze a number entered by the user.
Ask the user to enter a number.
Your program should determine:
- Whether the number is positive, negative, or zero.
- If it is positive or negative, also determine whether it is even or odd.

For example:

Enter number: -7

Negative

Odd

```python
n = int(input("Enter a number:- "))
if n > 0 and n % 2 == 0:
    print(f"{n} is even number, and a positive number.")
elif n > 0 and n % 2 != 0:
    print(f"{n} is odd number, and a positive number.")
elif n < 0 and n % 2 == 0:
    print(f"{n} is even number, and a negative number.")
elif n < 0 and n % 2 != 0:
    print(f"{n} is odd number, and a negative number.")
else:
    print("Zero")
```

##### Question 8: ATM Withdrawal
You are developing an ATM withdrawal system.
The user's account balance is:
50000
Ask the user to enter the withdrawal amount.
Apply these rules:
- If the amount is less than or equal to 0, display "Invalid Amount".
- If the amount is greater than the balance, display "Insufficient Balance".
- Otherwise, process the withdrawal and display the remaining balance.

```python
amountBalance = 50000
withdraw = int(input("Enter amount:- "))
if withdraw > amountBalance:
    print("Insufficient Balance.")
elif withdraw <= 0:
    print("Invalid Amount.")
else:
    remainingBalance = amountBalance - withdraw
    print(f"Remaining Balance:- {remainingBalance}")
```

- Have at least 50% in the entry test.

##### Question 12: Employee Performance & Promotion System

Promotion Rules

An employee can be promoted only if:

Performance >= 90

AND Experience >= 5

→ Senior Promotion

Performance >= 80

AND Experience >= 3

Otherwise

→ Junior Promotion

Bonus

Calculate the bonus category:

Performance >= 80

→ 15% bonus

Performance >= 70

Performance: 87
